```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)
```

```python
text_columns = df.select_dtypes(include="object").columns

text_columns
```

```python
for col in text_columns:
    df[col] = df[col].str.strip()
```

```python
yes_no_columns = [
    "private_office_available",
    "dedicated_desk",
    "hot_desk",
    "virtual_office",
    "meeting_room",
    "conference_room",
    "wifi",
    "parking",
    "power_backup",
    "cafeteria",
    "housekeeping",
    "security",
    "24x7_access",
    "air_conditioning",
    "printing",
    "event_space",
    "community_events",
    "managed_office",
    "build_to_suit",
    "enterprise_solution"
]
```

```python
for col in yes_no_columns:
    df[col] = (
        df[col]
        .astype("string")
        .str.strip()
        .str.lower()
        .map({
            "yes": "Yes",
            "y": "Yes",
            "no": "No",
            "n": "No"
        })
    )
```

```python
df["wifi"].value_counts(dropna=False)
```

```python
price_columns = [
    "price_per_seat_month",
    "day_pass_price",
    "meeting_room_price_hour"
]
```

```python
for col in price_columns:
    df[col] = (
        df[col]
        .astype("string")
        .str.replace("₹", "", regex=False)
        .str.replace(",", "", regex=False)
        .str.replace("/month", "", regex=False)
        .str.replace("+ GST", "", regex=False)
        .str.strip()
    )

    df[col] = pd.to_numeric(
        df[col],
        errors="coerce"
    )
```
